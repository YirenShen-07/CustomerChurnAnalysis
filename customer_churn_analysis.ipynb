# Customer Churn Analysis

A reproducible comparison of a majority-class baseline, Logistic Regression and Logistic GAM on the public IBM Telco benchmark. Positive class: Churn=Yes.

The 80/20 split is stratified. All imputers, scaling and categorical encoding are fitted only on training rows. GAM uses three numeric spline terms and categorical factors; plots use the actual transformed feature order. Linear Regression is retained only as an exploratory continuous-output comparison.
